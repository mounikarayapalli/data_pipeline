```python
# Import Pandas for data manipulation and analysis
import pandas as pandas

# Import Seaborn to access the built-in Titanic dataset
import seaborn as sns

# Load the Titanic dataset from Seaborn
df = sns.load_dataset('titanic')

# Display the total number of rows and columns in the dataset
print(f"Loaded {len(df)} rows and {len(df.columns)} columns from the Titanic Dataset.")

# Display the first five rows of the dataset to understand its structure
df.head()
```

```python
# Display all column names present in the dataset
display(df.columns)

# Display dataset information, including column names, data types and non-null values
df.info()

# Display descriptive statistical information such as count, mean, standard deviation, minimum and maximum values
display(df.describe())
```

```python
# Check the number of missing (null) values in each column of the dataset
df.isnull().sum()
```

```python
# Analysis check distribution of dataset
print(df['sex'].value_counts())
print(df['embarked'].value_counts())
print(df['survived'].value_counts())
```

```python
# Replace missing values in the age column with the median age
df['age'] = df['age'].fillna(df['age'].median())

# Replace missing values in the embarked column with its most frequently occurring value (mode)
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

# Remove the deck column because it contains many missing values
df = df.drop(columns=['deck'])

# Remove the embark_town column because the embarked column already contains similar information
df = df.drop(columns=['embark_town'])

# Check the number of remaining missing values in each column after cleaning
df.isnull().sum()
```

```python
#Categorical Encoding
df['sex']=df['sex'].map({'male':0,'female':1})
df[['sex']].head()
```

```python
#Validation 
assert df.isnull().sum().sum() == 0,"There are missing values in dataset."
assert df['sex'].isin([0,1]).all(),"There are unexpected Val in 'sex' col"
print("Validation passed: No missing val and 'sex' col is correctly encoded")
df.info()
```
